# Getting Started with Neural Networks

If you are unfamiliar with neural networks and CNNs, we have made a notebook for you to get started: **Intro to neural networks with Duckietown**. **This notebook is completely optional.** You can run it in one of two ways.

### Option 1 (recommended): in this editor

Open [intro_to_nn.ipynb](./intro_to_nn.ipynb) (in this same folder) and run its cells in order. PyTorch and torchvision are already installed in the editor, so there is nothing to install and you do not need a GPU. The notebook downloads the small CIFAR-10 dataset (about 170 MB) to `assets/data/cifar10/`.

### Option 2: on Google Colab

To access the [Colab version of the notebook](https://colab.research.google.com/github/duckietown/lx-recipe-object-detection/blob/ente/assets/colab/dt_intro_nn_with_pytorch.ipynb):
1) You should first click on the link to open up the notebook on Colab. 
2) Then you should click on the "Copy to Drive" button at the top left.
3) In order to get access to the GPU resources, you should now click on the "Connect" button at the top right. Afterwards you should see that you have been allocated RAM and disk.
4) Complete the steps to define and evaluate your own Neural Net using PyTorch.

Once you have completed the notebook (if you felt the need to do so), you can move on to the second notebook [02-Setup-and-Data-Collection](../02-Setup-Data-Collection/setup.ipynb).
